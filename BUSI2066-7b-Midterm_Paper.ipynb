## Q1. Measuring customer loyalty — 10 points

The tea chain wants to measure **customer loyalty** in its next survey.

**(a)** A manager suggests measuring customer loyalty by **how much a customer spent on their most recent
visit**. Give one reason why this could be misleading. Suggest a better survey question and explain why it
measures loyalty more directly. (3 points)

**(b)** The draft survey asks:

> How old are you? ☐ Under 20 ☐ 20–30 ☐ 30–40 ☐ Over 40

Give **two** problems with this question, then rewrite it. (4 points)

**(c)** The draft also asks: *“How satisfied are you with our drinks and our staff? (1 = very dissatisfied,
5 = very satisfied)”*. What is wrong with this question, and how would you fix it? (3 points)

*An ideal answer should be less than 150 words in total. This is a guideline, not a strict word limit.*

✍️ **Your answer**

(a) Spending on the most recent visit is not a good loyalty measure because one-off purchases can be unusually large. A better question is: “How often have you visited our tea shop in the last 3 months?” Repeat visits measure loyalty more directly.

(b) The categories overlap at 30, and the age groups are not clearly defined or complete. A better version is: “What is your age? ___ years” or “Which age group are you in? Under 20; 20–29; 30–39; 40–49; 50 or over?”

(c) The question is double-barrelled because it mixes drinks and staff. Split it into two separate 1–5 questions, one for drinks and one for staff.

## Q2. Two ways to collect a sample — 10 points

The chain wants to know what **all** its customers think. It has two plans.

- **Plan 1.** A QR code is printed on every receipt. Anyone who wants to can scan it and answer. There is no reward.
- **Plan 2.** At each branch, at random times over one week, a staff member asks **every 10th customer** at
  the counter to answer five questions on a tablet.

**(a)** Which kind of bias is Plan 1 most open to? Which customers are likely to be **too many** in its sample? (4 points)

**(b)** Which plan gives a sample closer to a random sample, and why? Name **one** group of the chain's
customers that Plan 2 still misses. (3 points)

**(c)** Plan 1 gets 3,000 answers and Plan 2 gets 400. The manager prefers Plan 1 “because it is bigger”.
Reply to the manager in two or three sentences, using the idea of **population validity**. (3 points)

*An ideal answer should be less than 150 words in total. This is a guideline, not a strict word limit.*

✍️ **Your answer**

(a) Plan 1 is open to voluntary-response bias. Very satisfied and very dissatisfied people are more likely to answer, so they are overrepresented.

(b) Plan 2 is closer to a random sample because it uses systematic selection at random times. One group it may still miss is customers who visit outside the sampled times.

(c) A larger sample is not automatically more valid if the sample is biased. Population validity matters more than sample size, so a smaller, more representative sample can be better than a large voluntary one.

### 1h · Code and written · 2 points

Now turn the levels into **scores 1 to 5**. Report the **mean score** (2 decimal places) and the **median
level**. In one or two sentences: which of the two is the safer summary for this column, and why?

```python
score_map = {
    'very dissatisfied': 1,
    'dissatisfied': 2,
    'neutral': 3,
    'satisfied': 4,
    'very satisfied': 5,
}
scores = df['satisfaction'].map(score_map)
print('Mean score:', round(scores.mean(), 2))
print('Median level:', scores.median())
```

## Question 2 · Spend per visit — 20 points

### 2a · Code · 3 points

Describe `spend`. Report the **mean, median, standard deviation, first quartile, third
quartile, IQR** and the **upper outlier fence** (Q3 + 1.5 × IQR), each to 2 decimal places.

```python
spend = df['spend']
q1 = spend.quantile(0.25)
q3 = spend.quantile(0.75)
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr
summary = {
    'mean': spend.mean(),
    'median': spend.median(),
    'std': spend.std(ddof=1),
    'Q1': q1,
    'Q3': q3,
    'IQR': iqr,
    'Upper fence': upper_fence,
}
print({k: round(v, 2) for k, v in summary.items()})
```

### 2b · Written · 2 points

## Question 3 · From 420 customers to all customers — 20 points

Treat the 420 answers as a **random sample** of all the chain's customers. Use `spend` throughout.

### 3a · Code · 5 points

Estimate a **95% confidence interval** for the average spend of all customers **by formula**
— no bootstrap, no t-test. Show each step, each to 2 decimal places: *n*, the mean, the standard deviation
*s*, the standard error *s* / √*n*, and the interval **mean ± 2 × SE**.

```python
spend = df['spend']
n = len(spend)
mean_spend = spend.mean()
s = spend.std(ddof=1)
se = s / np.sqrt(n)
ci_low = mean_spend - 2 * se
ci_high = mean_spend + 2 * se

print(f'n = {n}')
print(f'mean = {mean_spend:.2f}')
print(f's = {s:.2f}')
print(f'SE = {se:.2f}')
print(f'95% CI = ({ci_low:.2f}, {ci_high:.2f})')
```

### 3b · Code · 2 points

Now estimate the same interval by **bootstrapping**, with **10,000 resamples** and
**`rng=2026`**. Report the standard error and the 95% interval.

```python
rng = np.random.default_rng(2026)
samples = df['spend'].to_numpy()
boot_means = []
for _ in range(10000):
    resample = rng.choice(samples, size=len(samples), replace=True)
    boot_means.append(resample.mean())
boot_means = np.array(boot_means)
print('Bootstrap SE:', round(boot_means.std(ddof=1), 2))
print('95% bootstrap interval:', np.quantile(boot_means, [0.025, 0.975]).round(2))
```

### 3c · Written · 3 points

In one sentence, compare it with question 3a. Then write **one sentence a manager could read**, saying what the
interval means.

✍️ **Your written answer:**

The bootstrap interval is very close to the formula-based interval, and it means we are 95% confident that the true mean spend per visit for all customers is between HK$59.06 and HK$65.67.

### 3e · Code · 2 points

Run a one-sample t-test, and report **t**, the **p-value** and the **95% CI**.

```python
mean_spend = df['spend'].mean()
se = df['spend'].std(ddof=1) / np.sqrt(len(df['spend']))
t_stat, p_val = stats.ttest_1samp(df['spend'], 60)
crit = stats.t.ppf(0.975, len(df['spend']) - 1)
ci_low = mean_spend - crit * se
ci_high = mean_spend + crit * se

print(f't = {t_stat:.2f}')
print(f'p-value = {p_val:.3f}')
print(f'95% CI = ({ci_low:.2f}, {ci_high:.2f})')
```
